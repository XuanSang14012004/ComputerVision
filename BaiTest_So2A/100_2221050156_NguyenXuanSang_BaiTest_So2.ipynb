```python
#Import các thư viện sử dụng
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
#Bài 1:
# Đọc ảnh
img=cv2.imread('pic/pic1.jpeg',0)

#hiển thị ảnh:
plt.imshow(img,cmap='gray')
plt.title('Ảnh gốc ')
plt.show()
```

```python
#độ phân giải của ảnh
h=img.shape[0]
w=img.shape[1]
print('Đô phân giải của ảnh :',h ,'x',w)
print('Số pixel điểm ảnh:',h*w)
```

```python

#Thực hiện Đảo ảnh: 
img_dao =img.max()-img


#Hiển thị kết quả ảnh trước và sau khi đảo ảnh:
plt. figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('ảnh gốc')

plt.subplot(1,2,2)
plt.imshow(img_dao,cmap='gray')
plt.title('ảnh đã đảo')
plt.show()

#lưu ảnh
cv2.imwrite('Pic/Saves/2221050156_neck_inv.jpg',img_dao)
```

```python
#Thực hiện biến đổi ảnh với hàm mũ:
c=14
a=1.5
img_gama= c * pow(img,a)
img_gama=img_gama.astype(np.int64) #ép kiểu về số nguyên



#Hiển thị kết quả trước và sau khi biến đổi:
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('Ảnh gốc ')

plt.subplot(1,2,2)
plt.imshow(img_gama,cmap='gray')
plt.title('Ảnh đã xử lý bằng hàm mũ (c=14 , a=1,5)')
plt.show()

#lưu ảnh
cv2.imwrite('Pic/Saves/2221050156_neck_gamma.jpg',img_gama)
```

```python
#Yêu cầu	1.4:	
#Khởi tạo kernel:
kernel = np.array([[0,0,-1,0,0],
                          [0,-1,-2,-1,0],
                          [-1,-2,16,-2,-1],
                           [0,-1,-2,-1,0],
                          [0,0,-1,0,0]
                           ])

print('Mặt nạ lọc identity:\n',kernel)
```

```python
#Thực hiện lọc ảnh với kernel ở trên:
img_kernel=cv2.filter2D(img_dao,-1,kernel)

#hiển thị kết quả:
plt.figure(figsize=(10,7))
plt.subplot(1,2,1)
plt.imshow(img_dao,cmap='gray')
plt.title('Ảnh đảo')

plt.subplot(1,2,2)
plt.imshow(img_kernel,cmap='gray')
plt.title('Ảnh qua bộ lọc kernel')

plt.show()

#lưu ảnh
cv2.imwrite('pic/Saves/2221050156_neck2D.jpg',img_kernel)
```

```python
#Bài 2:
#Yêu cầu 2.1:
# Đọc ảnh
img2=cv2.imread('pic/pic2.jpg',0)

#độ phân giải của ảnh
h=img2.shape[0]
w=img2.shape[1]
print('Đô phân giải của ảnh :',h ,'x',w)
print('Số pixel điểm ảnh:',h*w)

#Hiển thị ảnh và histogram của ảnh:
plt.figure(figsize=(6,7))
plt.subplot(2,1,1)
plt.imshow(img2,cmap='gray')
plt.title('Ảnh xám')

plt.subplot(2,1,2)
plt.hist(img2.ravel(),256)
plt.title('Biêu đồ histogram của ảnh')
plt.show()
```

```python
#Yêu cầu2.2:
#Thực hiện cân bằng sáng với ảnh Tối:

img2_cb=cv2.equalizeHist(img2)
#hiển thị 
plt.figure(figsize=(12,8))
plt.subplot(1,2,1)
plt.imshow(img2,cmap='gray')
plt.title('Ảnh gốc')

plt.subplot(1,2,2)
plt.imshow(img2_cb,cmap='gray')
plt.title('Ảnh đã cân bằng sáng Histogram của ảnh ')

plt.show()

#lưu ảnh
cv2.imwrite('pic/Saves/2221050156_Equalization1.png',img2_cb)
```

```python
#yêu cầu 2.3:
#️ Tạo đối tượng CLAHE
clahe = cv2.createCLAHE(clipLimit=14, tileGridSize=(30, 30))

# 2 Thực hiện cân bằng sáng cục bộ
img2_clahe = clahe.apply(img2_cb)

#  Hiển thị kết quả
plt.figure(figsize=(12,8))

plt.subplot(1,2,1)
plt.imshow(img2_clahe, cmap='gray')
plt.title('Ảnh đã cân bằng sáng theo vùng (30x30)')

plt.subplot(1,2,2)
plt.hist(img2_clahe.ravel(),256)
plt.title('Biêu đồ histogram của ảnh')

plt.show()

#lưu ảnh
cv2.imwrite('pic/Saves/2221050156_Equalization2.png',img2_clahe)
```

```python
#Bài 3:
#Yêu cầu 3.1:
# Đọc ảnh
img3=cv2.imread('Pic/Bai3/pic3_01.jpg', cv2.IMREAD_COLOR_RGB)

#độ phân giải của ảnh
h=img3.shape[0]
w=img3.shape[1]
print('Đô phân giải của ảnh :',h ,'x',w)
print('Số pixel điểm ảnh:',h*w)

#Tách các kênh màu B, G, R từ ảnh màu
b_channel = None
g_channel = None
r_channel = None

#LẤY DỮ LIỆU CÁC KÊNH MÀU:BGR (B=index:0|G=index:1|R=index:2)
b_channel = img3.copy()
b_channel[:, :, [1,2]] = 0 #Chuyển đổi kênh GR = 0 giữ lại Kênh B

g_channel = img3.copy()
g_channel[:, :, [0, 2]] = 0 #Chuyển đổi kênh BR=0 giữ lại kênh G

r_channel = img3.copy()
r_channel[:, :, [0, 1]] = 0 #Chuyển đổi kênh BG=0 giữ lại Kênh R


#Hiển thị ảnh gốc và các kênh màu RGB với Matplotlib

plt.figure(figsize=(10,8))
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img3, cv2.COLOR_BGR2RGB))
plt.title('Ảnh RGB')

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(b_channel, cv2.COLOR_BGR2RGB))
plt.title('Blue Channel')

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(g_channel, cv2.COLOR_BGR2RGB))
plt.title('Green Channel')

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(r_channel, cv2.COLOR_BGR2RGB))
plt.title('Red Channel')
plt.show()
```

```python
#Yêu cầu 3.2:
# 1 Chuyển ảnh sang xám
img_gray = cv2.cvtColor(img3,cv2.COLOR_BGR2GRAY)

#  Chuyển ảnh sang đen trắng (nhị phân)
_, img_binary = cv2.threshold(img_gray,127, 255,cv2.THRESH_BINARY)

#hiển thị 
plt.figure(figsize=(12,8))
plt.subplot(1,3,1)
plt.imshow(img3)
plt.title('Ảnh màu')


plt.subplot(1,3,2)
plt.imshow(img_gray,cmap='gray')
plt.title('Ảnh xám')

plt.subplot(1,3,3)
plt.imshow(img_binary,cmap='gray')
plt.title('Ảnh đen trắng ')

plt.show()

#lưu ảnh
cv2.imwrite('pic/Saves/2221050156_gray.png',img_gray)

cv2.imwrite('pic/Saves/2221050156_binary.png',img_binary)
```

```python
#Yêu cầu 3.3: 14/1
#Làm mịn ảnh với các kernel khác nhau:
img_blur_1 = cv2.blur(img3, (1,1))
img_blur_15 = cv2.blur(img3, (15,15))

# Hiển thị kết quả
plt.figure(figsize=(12,8))

plt.subplot(1,3,1)
plt.imshow(img3)
plt.title('Ảnh gốc')

plt.subplot(1,3,2)
plt.imshow(img_blur_1)
plt.title('Làm mờ trung bình (1,1)')

plt.subplot(1,3,3)
plt.imshow(img_blur_15)
plt.title('Làm mờ trung bình (15,15)')

plt.show()


```
