```python
#Import các thư viện sử dụng
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
#yêu cầu 1:
#đọc ảnh
img=cv2.imread('Pic/Pic_original/Test_01.jpg',cv2.IMREAD_COLOR_RGB)
#hiển thị ảnh
plt.imshow(img)
plt.show()
#độ phân giải của ảnh
h=img.shape[0]
w=img.shape[1]
print('Đô phân giải của ảnh :',h ,'x',w)
print('Số pixel điểm ảnh:',h*w)
```

```python
#YÊU CẦU 2
#chuyển sang hệ màu YCrCb,LAB.
img_YCrCb=cv2.cvtColor(img,cv2.COLOR_BGR2YCrCb)
plt.subplot(1,2,1)
plt.imshow(img_YCrCb)
plt.title('Ảnh hệ màu YCrCb')

img_lab=cv2.cvtColor(img,cv2.COLOR_BGR2LAB)
plt.subplot(1,2,2)
plt.imshow(img_lab)
plt.title('Ảnh hệ màu LAB')

plt.show()
cv2.imwrite('Pic/Saves/2221050156_YCrCb.jpg',img_YCrCb)
cv2.imwrite('Pic/Saves/2221050156_LAB.jpg',img_lab)
```

```python
# YÊU CẦU 3:
# nhập kích thước bạn muốn thay đổi
scale=int(input('Nhập vào tỉ lệ % muốn thay đổi kích thước ảnh:'))
#tính tỉ lệ ảnh mới theo người dùng nhập
h_new=int(h*scale/100)
w_new=int(w*scale/100)

std_old='Kích thước ảnh gốc :'+str(h)+'x'+str(w)
std_new='Ảnh đã Scale:'+str(h_new)+'x'+str(w_new)

# Chọn thuật toán nội suy
if scale < 100:
    img_scale=cv2.resize(img,(w_new,h_new),cv2.INTER_LANCZOS4)
    filename = "Pic/Saves/2221050156_Scale_In.jpg"   # Ảnh thu nhỏ
else:
     img_scale=cv2.resize(img,(w_new,h_new),cv2.INTER_AREA)
     filename = "Pic/Saves/2221050156_Scale_Out.jpg"  # Ảnh phóng to
    
#hiển thị kết quả ảnh gốc - ảnh sau khi thay đổi kích thước 
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title(std_old)

plt.subplot(1,2,2)
plt.imshow(img_scale)
plt.title(std_new)
plt.show()
#lưu ảnh 
cv2.imwrite(filename,cv2.cvtColor(img_scale,cv2.COLOR_BGR2RGB))
```

```python
#YÊU CẦU 4:
#hiển thị ở chế độ ảnh  xám
img_gray=cv2.imread('Pic/Pic_original/Test_01.jpg',0)
#hiển thị
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.imshow(img_gray,cmap='gray')
plt.title('Ảnh gray')
#tăng mật độ của lưới :
plt.xticks(range(0,img_gray.shape[1],100),rotation=90)
plt.yticks(range(0,img_gray.shape[0],100))
plt.grid(ls=':')#hiển thị lưới

img_gray[100:300,200:400]=255
img_gray[500:700,200:400]=0
img_gray[100:300,1000:1200]=0
img_gray[500:700,1000:1200]=255
plt.subplot(1,2,2)
plt.imshow(img_gray,cmap='gray')
plt.title('Ảnh đã xử lý')

plt.show()
#lưu ảnh
cv2.imwrite('Pic/Saves/2221050156_img_gray.jpg',img_gray)
```

```python
#YÊU VẦU 5
#  Xoay ảnh 180 độ 
img_180 = cv2.rotate(img_gray, cv2.ROTATE_180)

#  Lật ảnh theo chiều ngang 
img_y = cv2.flip(img_180, 1)

# Hiển thị kết quả
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap="gray")
plt.title("Ảnh Gray Đã Xử Lý")

#plt.subplot(1,3,2)
#plt.imshow(img_180, cmap="gray")
#plt.title("Ảnh Xoay 180°" )

plt.subplot(1,2,2)
plt.imshow(img_y, cmap="gray")
plt.title("Ảnh Đã Xoay 180- LẬT NGANG")

plt.show()
#lưu ảnh
cv2.imwrite('Pic/Saves/2221050156_Rotate_Flip.jpg',img_y)
```

```python
#YÊU CẦU 7
import cv2
import os
import matplotlib.pyplot as plt

# Thông tin sinh viên
MaSV = "2221050156"
ngay_sinh = int(input("Nhập ngày sinh của bạn (1-31): "))

# Thư mục ảnh
input_dir = "Pic/Pic_original"
output_dir = "Pic/Pic_New"
#os.makedirs(output_dir, exist_ok=True)  # tạo thư mục nếu chưa có

# Lấy danh sách file ảnh
files = [f for f in os.listdir(input_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

# Xử lý và lưu ảnh
for i, file in enumerate(files, start=1):
    path = os.path.join(input_dir, file)
    img_7 = cv2.imread(path)

    if img_7 is None:
        print(f" Không đọc được ảnh: {file}")
        continue

    # Resize về 128x96
    img_resized = cv2.resize(img_7, (128, 96))

    # Lật ảnh theo ngày sinh
    if ngay_sinh <= 15:
        img_flipped = cv2.flip(img_resized, 0)   # lật theo trục X
    else:
        img_flipped = cv2.flip(img_resized, 1)   # lật theo trục Y

    # Lưu ảnh theo quy tắc MaSV_i.png
    save_name = f"{MaSV}_{i}.png"
    save_path = os.path.join(output_dir, save_name)
    cv2.imwrite(save_path, img_flipped)

print("Đã xử lý và lưu toàn bộ ảnh vào thư mục Pic_New.")

# Hiển thị toàn bộ ảnh đã xử lý
# Hiển thị ảnh 4 hàng
processed_files = [f for f in os.listdir(output_dir) if f.endswith(".png")]
num_files = len(processed_files)
num_rows = 4
num_cols = math.ceil(num_files / num_rows)  # tính số cột cần thiết

plt.figure(figsize=(20, 12))  # figure lớn hơn

for idx, file in enumerate(processed_files, start=1):
    path = os.path.join(output_dir, file)
    img = cv2.imread(path)
    if img is None:
        continue

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.subplot(num_rows, num_cols, idx)  # đặt vào lưới đúng
    plt.imshow(img_rgb)
    plt.title(file)
    plt.axis("off")

plt.show()
```
