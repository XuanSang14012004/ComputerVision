# 2. Phát hiện chuyển động (Motion Detection)
---

Phát hiện chuyển động là bài toán trong xử lý video nhằm xác định các vùng ảnh có sự thay đổi theo thời gian, biểu hiện cho sự dịch chuyển vị trí của vật thể trong không gian quan sát.


```python
#=================================================
# PHƯƠNG PHÁP TRỪ NỀN (BACKGROUND SUBTRACTION):
#================================================
import cv2

# ------ 1. Mở nguồn video --------
# Mở camera
#cap = cv2.VideoCapture(1)

# Hoặc dùng file video:
#cap = cv2.VideoCapture("Video/slow_traffic_small.mp4")
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print("Không mở được video!")
    raise SystemExit

# --- 2. Khởi tạo bộ trừ nền MOG2 ---
# history: số frame dùng để học nền
# varThreshold: ngưỡng phân tách nền / tiền cảnh (motion)
# detectShadows=True: đánh dấu bóng bằng màu xám đậm (giá trị ~127)
fgbg = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=16,
    detectShadows=True
)


while True:
    ret, frame = cap.read()
    if not ret:
        break

    # --- 3. Áp dụng background subtraction ---
    fgmask = fgbg.apply(frame)  # foreground mask (vùng chuyển động)

    # --- 4. Tiền xử lý mask để giảm nhiễu ---
    # Bỏ bóng: coi pixel xám (127) là nền
    _, fgmask_bin = cv2.threshold(fgmask, 200, 255, cv2.THRESH_BINARY)

    # Giãn (dilate) để lấp lỗ nhỏ, gộp vùng chuyển động
    fgmask_bin = cv2.dilate(fgmask_bin, None, iterations=2)

    # --- 5. Tìm contour trên vùng chuyển động ---
    contours, _ = cv2.findContours(
        fgmask_bin.copy(),
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    for cnt in contours:
        # Bỏ qua vùng quá nhỏ (nhiễu)
        if cv2.contourArea(cnt) < 500:
            continue
        x, y, w, h = cv2.boundingRect(cnt)
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)

    cv2.putText(frame, "Motion Detection - Background Subtraction", (10, 20),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

    # --- 6. Hiển thị kết quả---
    cv2.imshow("Frame", frame)           # khung hình gốc + box
#    cv2.imshow("Foreground Mask (raw)", fgmask)  # mask gốc từ MOG2
    cv2.imshow("Foreground Mask (clean)", fgmask_bin)  # mask đã lọc

    if cv2.waitKey(30) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
#================================================================
# 2. PHƯƠNG PHÁP LUỒNG QUANG HỌC (Optical Flow - Lucas–Kanade)
#===============================================================
import cv2
import numpy as np

# ------ 1. Mở nguồn video --------
# Mở camera:
# cap = cv2.VideoCapture(0)

# Hoặc dùng file video:
cap = cv2.VideoCapture("Video/slow_traffic_small.mp4")

if not cap.isOpened():
    print("Không mở được video!")
    raise SystemExit

cv2.startWindowThread()

# ------ 2. Tham số Shi–Tomasi corner detection (tìm điểm đặc trưng ban đầu) ------
st_params  = dict(maxCorners = 100,qualityLevel = 0.3,minDistance = 7,blockSize = 7
)

# ------ 3. Tham số Lucas–Kanade Optical Flow ------
lk_params = dict(winSize  = (15, 15),maxLevel = 2,
    criteria = (cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT,10,0.03))

# Tạo bảng màu ngẫu nhiên cho các điểm theo vết
color = np.random.randint(0, 255, (100, 3))

# ------ 4. Đọc frame đầu tiên và tìm điểm đặc trưng ------
ret, old_frame = cap.read()
if not ret:
    print("Không đọc được frame đầu tiên!")
    cap.release()
    raise SystemExit

old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)

# Tìm các điểm góc (corners) để theo vết
p0 = cv2.goodFeaturesToTrack(old_gray, mask=None, **st_params)

# Tạo mask để vẽ đường đi (trajectory)
mask = np.zeros_like(old_frame)

# ------ 5. Vòng lặp xử lý từng frame ------
while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 5.1. Tính optical flow (Lucas–Kanade)
    if p0 is None:
        # Nếu không còn điểm nào để theo vết → tìm lại điểm mới
        p0 = cv2.goodFeaturesToTrack(old_gray, mask=None, **st_params)

    if p0 is None:
        # Vẫn không tìm được điểm → bỏ qua frame này
        cv2.imshow("Optical flow", frame)
    else:
        p1, st, err = cv2.calcOpticalFlowPyrLK(old_gray, frame_gray, p0, None, **lk_params)

        # 5.2. Lọc ra các điểm "tốt"
        if p1 is not None:
            good_new = p1[st == 1]
            good_old = p0[st == 1]

            # 5.3. Vẽ quỹ đạo chuyển động (đường + chấm)
            for i, (new, old) in enumerate(zip(good_new, good_old)):
                a, b = new.ravel()
                c, d = old.ravel()
                mask = cv2.line(mask, (int(a), int(b)), (int(c), int(d)),color[i].tolist(), 2)
                frame = cv2.circle(frame, (int(a), int(b)), 5,color[i].tolist(), -1)

            img = cv2.add(frame, mask)
            cv2.putText(img, "Optical Flow - Lucas-Kanade",(10, 30),cv2.FONT_HERSHEY_SIMPLEX,1, (0, 0, 255), 2)
            
            cv2.imshow("Optical flow", img)

            # Cập nhật frame và điểm cho vòng sau
            old_gray = frame_gray.copy()
            p0 = good_new.reshape(-1, 1, 2)
        else:
            cv2.imshow("Optical flow", frame)
            old_gray = frame_gray.copy()
            p0 = None

    if cv2.waitKey(30) & 0xFF == ord("q"):
        break

# Giải phóng bộ nhớ:
cap.release()
cv2.destroyAllWindows()
```

# 3. Theo vết đối tượng (Object tracking)
---
Theo vết đối tượng (Object Tracking) là bài toán trong thị giác máy tính nhằm xác định vị trí liên tục của một hay nhiều đối tượng theo thời gian trong chuỗi khung hình video sau khi đối tượng đã được phát hiện hoặc chỉ định ban đầu.

Mục tiêu của Object tracking là xác định vị trí của một hoặc nhiều đối tượng (target) trong các frame liên tiếp trong Video

* Single Object tracking (SOT) – Chỉ có một object được theo dõi ngay cả khi môi trường có nhiều  đối tượng.

* Muiltiple Object tracking (OT) – Nhiều đối tượng được theo dõi theo thời gian, thậm chí nó có thể theo dõi các đối tượng mới xuất hiện trong video.


### a. Phương pháp MeanShift
---
MeanShift là phương pháp theo vết đối tượng dựa trên phân bố màu sắc của vùng đối tượng, sử dụng thuật toán dịch chuyển trung bình (mean shift) để tìm cực đại mật độ xác suất trong không gian histogram màu.

**Nguyên lý:**
1. Trích histogram màu của ROI ban đầu.
2. Tính back-projection trên frame mới → tạo bản đồ xác suất vị trí đối tượng.
3. Dịch chuyển cửa sổ ROI dần về vị trí có mật độ xác suất cao nhất (cực đại).


```python
import cv2
import numpy as np

# ========================
# 1. Đọc video
# ========================
cap = cv2.VideoCapture("Video/slow_traffic_small.mp4")

if not cap.isOpened():
    print("Không mở được video!")
    raise SystemExit

# ========================
# 2. Đọc frame đầu tiên
# ========================
ret, frame = cap.read()
if not ret:
    print("Không đọc được frame đầu tiên!")
    cap.release()
    raise SystemExit

# ================================
# 3. ROI ban đầu (hard-code)
# ==============================
x, y, w, h = 300, 200, 100, 50   # TỌA ĐỘ DEMO

x, y, w, h = int(x), int(y), int(w), int(h)
track_window = (x, y, w, h)

# ==============================
# 4. Tạo histogram màu ROI
# ===============================
roi = frame[y:y+h, x:x+w]
hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

mask = cv2.inRange(
    hsv_roi,
    np.array((0., 60., 32.)),
    np.array((180., 255., 255.))
)

roi_hist = cv2.calcHist([hsv_roi], [0], mask, [180], [0, 180])
cv2.normalize(roi_hist, roi_hist, 0, 255, cv2.NORM_MINMAX)

# ========================
# 5. Tiêu chuẩn dừng MeanShift
# ========================
term_crit = (
    cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT,
    10,   # Số vòng lặp tối đa
    1     # Dịch chuyển < 1 pixel
)

# 6. Tracking loop
while True:
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Back-projection
    dst = cv2.calcBackProject([hsv], [0], roi_hist, [0, 180], 1)

    # Áp dụng MeanShift
    _, track_window = cv2.meanShift(dst, track_window, term_crit)

    # Lấy bounding box
    x, y, w, h = track_window

    # Vẽ box
    tracked_frame = cv2.rectangle(frame.copy(),(x, y), (x + w, y + h),(0, 255, 0),2)

    #Hiển thị kết quả:
    cv2.putText(tracked_frame,"MeanShift Tracking",(10, 30),cv2.FONT_HERSHEY_SIMPLEX,1,(0, 0, 255), 2)
    cv2.imshow("Example MeanShift", tracked_frame)

    if cv2.waitKey(100) & 0xFF == ord("q"):
        break

# Giải phóng bộ nhớ
cap.release()
cv2.destroyAllWindows()
```

## b. Phương pháp Camshift
---

CamShift là thuật toán Single Object Tracking dựa trên histogram màu, phát triển từ MeanShift, cho phép theo vết đối tượng liên tục trong video với khả năng tự điều chỉnh kích thước và hướng ROI theo sự thay đổi của đối tượng.

**Nguyên lý:**
1. Dựa trên phân bố màu sắc (HSV histogram) của ROI ban đầu.
2. Tính back-projection để xây dựng bản đồ xác suất đối tượng.
3. Tìm cực đại mật độ bằng MeanShift, sau đó:
    * Cập nhật tâm ROI
    * Tự điều chỉnh kích thước & hướng bounding box theo phân bố pixel.


```python
import cv2
import numpy as np

# ========================
# 1. Đọc video
# ========================
cap = cv2.VideoCapture("Video/slow_traffic_small.mp4")
# Nếu dùng webcam (macOS):
# cap = cv2.VideoCapture(0, cv2.CAP_AVFOUNDATION)

if not cap.isOpened():
    print("Không mở được video!")
    raise SystemExit

# ========================
# 2. Đọc frame đầu tiên
# ========================
ret, frame = cap.read()
if not ret:
    print("Không đọc được frame đầu tiên!")
    cap.release()
    raise SystemExit

# ========================
# 3. ROI ban đầu (hard-code demo)
# ========================
x, y, w, h = 300, 200, 100, 50    # toạ độ ví dụ
x, y, w, h = int(x), int(y), int(w), int(h)
track_window = (x, y, w, h)

# ========================
# 4. Tạo histogram màu cho ROI
# ========================
roi = frame[y:y+h, x:x+w]
hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

mask = cv2.inRange(
    hsv_roi,
    np.array((0., 60., 32.)),
    np.array((180., 255., 255.))
)

roi_hist = cv2.calcHist([hsv_roi], [0], mask, [180], [0, 180])
cv2.normalize(roi_hist, roi_hist, 0, 255, cv2.NORM_MINMAX)

# ========================
# 5. Tiêu chuẩn dừng cho CamShift
# ========================
term_crit = (cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT,10,1)

# ========================
# 6. Vòng lặp CamShift tracking
# ========================
while True:
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Back-projection theo histogram ROI
    dst = cv2.calcBackProject([hsv], [0], roi_hist, [0, 180], 1)

    # Áp dụng CamShift: trả về rotated rect + track_window mới
    ret_cs, track_window = cv2.CamShift(dst, track_window, term_crit)

    # Chuyển rotated rect thành 4 điểm đa giác
    pts = cv2.boxPoints(ret_cs)
    pts = pts.astype(int)

    img2 = frame.copy()
    img2 = cv2.polylines(img2, [pts], True, (0, 255, 0), 2)

    #Hiển thị kết quả:
    cv2.putText(img2, "CamShift Tracking",(10, 30),cv2.FONT_HERSHEY_SIMPLEX,1,(0, 0, 255),2)
    cv2.imshow("Example CamShift", img2)

    if cv2.waitKey(150) & 0xFF == ord("q"):
        break

# ========================
# 7. Giải phóng bộ nhớ
# ========================
cap.release()
cv2.destroyAllWindows()
```

# Sử dụng module Tracker trong OpenCV
---
OpenCV Tracker module cung cấp tập hợp các thuật toán theo vết đối tượng (Single Object Tracking) cho video, giúp:
* Theo dõi 1 đối tượng duy nhất sau khi đã xác định ROI ban đầu.
* Dự đoán vị trí bounding box qua các frame liên tiếp.
* Giữ ID đối tượng trong suốt quá trình.

```python
import cv2
import numpy as np

# =================================
# 1. Mở video source
# =================================
cap = cv2.VideoCapture('Video/crime-object.mp4')

if not cap.isOpened():
    print("Không mở được video nguồn!")
    raise SystemExit

# Thiết lập tham số lưu kết quả
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
size   = (width, height)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('Object_tracking.mp4',fourcc,20.0,size)

# ===================================
# 2. Chọn ROI ban đầu để theo vết
# ===================================
ret, frame = cap.read()
    
roi = cv2.selectROI("Select Object to Track",frame,showCrosshair=False)
cv2.destroyWindow("Select Object to Track")


# ==================================================
# 3. Khởi tạo bộ theo vết tracker của OpenCV
# =====================================================
tracker = cv2.TrackerCSRT_create() #Thay đổi ROI vùng theo vết
#tracker = cv2.TrackerMIL_create() #Cố định ROI vùng theo vết

#Sinh viên thử với các Tracker khác: Boosting, KCF, MOSSE, ....

roi = tuple(map(int, roi))
tracker.init(frame, roi)

# ============================================
# 4. Xử lý trên từng frame liên tiếp để xác vùng ROI phù hợp
# ==============================================
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Cập nhật vị trí ROI
    success, roi = tracker.update(frame)
    (x, y, w, h) = tuple(map(int, roi))

    # Xử lý kết quả cho các vùng ROI 
    if success:
        cv2.rectangle(frame,(x, y),(x + w, y + h),(0, 255, 0),3)
        cv2.putText(frame,"Object",(x, y - 10),cv2.FONT_HERSHEY_SIMPLEX,0.8,(0, 255, 0),2)

    else:
        cv2.putText(frame,"Fail to track the object",(150, 200),cv2.FONT_HERSHEY_SIMPLEX,1,(0, 0, 255),2)
    
    #Hiển thị kết quả theo vết & ghi video
    cv2.imshow("Demo Object Tracking", frame)
    out.write(frame)

    # Exit bằng phím q
    if cv2.waitKey(20) & 0xFF == ord("q"):
        break

#Giải phóng bộ nhớ
cap.release()
out.release()
cv2.destroyAllWindows()
```

```python
#======================================
#1. PHÁT HIỆN KHUÔN MẶT TRONG VIDEO
#========================================

#Kết nối tới nguồn Camera; Nếu máy tính không có Webcam SV sử dụng file video vtv.mp4
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Không mở được webcam.")
    exit()

# Tải bộ phát hiện khuôn mặt Haar Cascade
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

#Xử lý từng Frame trong video:
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Chuyển sang ảnh xám để detect ổn định hơn
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Phát hiện khuôn mặt:
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.3,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # Vẽ Bounding box và Text lên frame gốc
    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(frame, "Face",(x, y-10),cv2.FONT_HERSHEY_SIMPLEX,1.3,(0, 255, 0),5)

    #Hiển thị:
    cv2.imshow("Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
#=====================================================
#2. NHẬN DẠNG KHUÔN MẶT
#=====================================================
#Hàm trích xuất đặc trưng khuôn mặt đơn giản:
def extract_face_feature(gray_face, size=(100, 100)):
    # resize về kích thước cố định
    face_resized = cv2.resize(gray_face, size)
    # chuyển sang vector 1 chiều
    vec = face_resized.flatten().astype("float32")
    # chuẩn hóa L2
    norm = np.linalg.norm(vec) + 1e-8
    vec /= norm
    return vec
```

```python
#Chuẩn bị 1 ảnh mẫu person_a.jpg (mặt thẳng, rõ).
#Dùng Haar để tìm mặt trong ảnh mẫu → trích đặc trưng với hàm extract_face_feature xây dựng ở trên.
import cv2 
import numpy as np
import matplotlib.pyplot as plt

# Đọc ảnh mẫu người A
img = cv2.imread("Person_A.jpg")

if img is None:
    print("Không đọc được ảnh đầu vào.jpg")
    exit()

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Phát hiện khuôn mặt trong ảnh mẫu
face_cascade = cv2.CascadeClassifier( cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
ref_faces = face_cascade.detectMultiScale(
    img_gray, 
    scaleFactor=1.3, 
    minNeighbors=5, 
    minSize=(30, 30)
)

if len(ref_faces) == 0:
    print("Không phát hiện được mặt trong ảnh mẫu.")
    exit()

#Tách lấy khuôn mặt trong ảnh:
(x, y, w, h) = ref_faces[0]
face_roi = img_gray[y:y+h, x:x+w]

#Sử dụng hàm extract_face_feature để trích xuất đặc trưng:
ref_feature = extract_face_feature(face_roi)

print("Đã trích đặc trưng khuôn mặt người A.")
plt.imshow(face_roi,cmap='gray')
plt.show()
```

```python
#Nhận dạng khuôn mặt trong video với khuôn mặt trong ảnh đầu vào
# Thiết lập ngưỡng so sánh (có thể điều chỉnh)
THRESHOLD = 0.4

#Đọc dữ liệu từ nguồn Camera:
cap = cv2.VideoCapture(1, cv2.CAP_AVFOUNDATION)
if not cap.isOpened():
    print("Không mở được webcam.")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray, scaleFactor=1.3, minNeighbors=5, minSize=(30, 30)
    )

    for (x, y, w, h) in faces:
        face_roi = gray[y:y+h, x:x+w]
        cur_feature = extract_face_feature(face_roi)

        # khoảng cách Euclid giữa mặt hiện tại và mặt người A
        dist = np.linalg.norm(cur_feature - ref_feature)
        #Khoản cách dist càng ngắn 2 khuôn mặt càng giống nhau: 
        if dist < THRESHOLD:
            label = "Person A"
            color = (0, 255, 0)
        else:
            label = "Unknown"
            color = (0, 0, 255)

        cv2.rectangle(frame, (x, y), (x+w, y+h), color, 2)
        cv2.putText(frame, f"{label} ({dist:.2f})",(x, y-10),cv2.FONT_HERSHEY_SIMPLEX,1.3,color,5)

    cv2.imshow("Face Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

Sinh viên tìm hiểu phát hiện và nhận dạng khuôn mặt, thực hiện nhận dạng 3 khuôn mặt khác nhau?

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ================== HÀM TRÍCH ĐẶC TRƯNG LBP ==================
def extract_face_feature(face_gray, size=(100, 100)):
    face = cv2.resize(face_gray, size)
    lbp = np.zeros_like(face)

    for i in range(1, face.shape[0] - 1):
        for j in range(1, face.shape[1] - 1):
            center = face[i, j]
            binary = ''
            binary += '1' if face[i-1, j-1] > center else '0'
            binary += '1' if face[i-1, j]   > center else '0'
            binary += '1' if face[i-1, j+1] > center else '0'
            binary += '1' if face[i, j+1]   > center else '0'
            binary += '1' if face[i+1, j+1] > center else '0'
            binary += '1' if face[i+1, j]   > center else '0'
            binary += '1' if face[i+1, j-1] > center else '0'
            binary += '1' if face[i, j-1]   > center else '0'
            lbp[i, j] = int(binary, 2)

    hist = cv2.calcHist([lbp], [0], None, [256], [0, 256])
    hist = cv2.normalize(hist, hist).flatten()
    return hist

# ================== LOAD HAAR CASCADE ==================
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# ================== ĐỌC ẢNH MẪU ==================
def load_reference_face(image_path):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Không đọc được {image_path}")
        exit()

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
    

    if len(faces) == 0:
        print(f"Không phát hiện được mặt trong {image_path}")
        exit()

    x, y, w, h = faces[0]
    face_roi = gray[y:y+h, x:x+w]
    feature = extract_face_feature(face_roi)
    return feature

print("Đang load ảnh mẫu...")

ref_A = load_reference_face("images/Person_A.jpg")
ref_B = load_reference_face("images/Person_B.png")
ref_C = load_reference_face("images/Person_C.png")

ref_features = [ref_A, ref_B, ref_C]
ref_names = ["Person A", "Person B", "Person C"]

print("Đã load xong 3 khuôn mặt mẫu")

# ================== NHẬN DẠNG TỪ WEBCAM ==================
THRESHOLD = 1.2   # Ngưỡng LBP

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Không mở được webcam")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        face_roi = gray[y:y+h, x:x+w]
        cur_feature = extract_face_feature(face_roi)

        distances = [np.linalg.norm(cur_feature - ref) for ref in ref_features]
        min_dist = min(distances)
        min_index = distances.index(min_dist)

        if min_dist < THRESHOLD:
            label = ref_names[min_index]
            color = (0, 255, 0)
        else:
            label = "Unknown"
            color = (0, 0, 255)

        cv2.rectangle(frame, (x, y), (x+w, y+h), color, 2)
        cv2.putText(
            frame,
            f"{label} ({min_dist:.2f})",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            color,
            2
        )

    cv2.imshow("Face Recognition - 3 People (OpenCV)", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
