```python
#Import các thư viện sử dụng
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

## Bài 1:
### Yêu	cầu	1.1:
Sinh	viên	đọc	và	hiển	thị	ảnh		Pic/Pic_T1.png	ở	chế	độ	ảnh	màu	
RGB

```python
img = cv2.imread('Pic/Pic_T1.png', cv2.IMREAD_COLOR_RGB)

if img is None:
    print ('Lỗi đọc ảnh!!')
else:
    plt.imshow(img)
    plt.title('Ảnh gốc')
    plt.show()
```

### Yêu	cầu	1.2.	
Lấy	thông	số	của	ảnh,	thực	hiện	Hiển	thị	kích	thước	và	số	pixel	
điểm	ảnh:	

```python
# xác định độ phân giản
h = img.shape[0]
w = img.shape[1]
print('1.Kích thước của ảnh:',h,'x',w)
print('2.Số pixel điểm ảnh:',h*w,'pixel')
```

### Yêu	cầu	1.3.
Thay	đổi	kích	thước	ảnh	trên	về		ảnh	mới	có	kích	thước	
(400x400)

```python
#Resize ảnh với các thuật toán khác nhau:

img_400 = cv2.resize(img, (400,400), cv2.INTER_AREA)

#Hiển thị ảnh thu nhỏ:
plt.imshow(img_400)
plt.title('Ảnh kích thước 400x400')
plt.grid()
# xác định độ phân giản
h = img_400.shape[0]
w = img_400.shape[1]
print('1.Kích thước của ảnh:',h,'x',w)
print('2.Số pixel điểm ảnh:',h*w,'pixel')

plt.show()
```

### Yêu	cầu	1.4.	
Chuyển	đổi	ảnh	màu	RGB	400x400	ở	trên	sang	ảnh	Đen	Trắng.	Sau	
đó	sử	dụng	phương	thức	Laplacian	để	xác	định	biên	của	bông	hoa,	thể	hiện	kết	
quả	như		hình	minh	họa.	Lưu	ảnh	biên	vào	thư	mục	Pic	đặt	tên:	MSV_Edge.jpg

```python
# 1 Chuyển ảnh sang xám
img_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

#  Chuyển ảnh sang đen trắng (nhị phân)
_, img_binary = cv2.threshold(img_gray,127, 255,cv2.THRESH_BINARY)
img_lap = cv2.Laplacian(img_binary, -1, 3)  
#hiển thị 
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.imshow(img_binary,cmap='gray')
plt.title('Ảnh ở chế độ đen - trắng ')
plt.subplot(1,2,2)    
plt.imshow(img_lap, cmap='gray')
plt.title('Đường biên của cánh hoa theo Lapacian')
plt.show()

cv2.imwrite('Pic/2221050156_Edge.png',img_lap)
```

## Bài 2:

Sinh	viên	đọc	ảnh	tầm

```python
#Đọc ảnh ở chế độ ảnh xám:
img_2 = cv2.imread('Pic/Pic_T2.jpeg',0)
h = img_2.shape[0]
w = img_2.shape[1]
# hiển thị ảnh gốc 
if img_2 is None:
    print('Lỗi đọc ảnh')
else:
    plt.imshow(img_2,cmap='gray')
    plt.xticks(range(0,img_2.shape[1],25))
    plt.grid(ls='--') 
    plt.title(f'ẢNH GỐC BAN ĐẦU - KÍCH THƯỚC :{h} x {w} pixel')
    plt.show()
```

### Yêu	cầu	2.2.
Sử	dụng	phương	pháp	
cắt	ngưỡng	ảnh	–	threshold	với	loại	
cắt	ngưỡng	(type)	là	
THRESH_BINARY_INV	và	các	giá	trị	
ngưỡng	phù	hợp	để	thu	được	ảnh	
như	sau:	

```python
T = 25
T, pic_inv = cv2.threshold(img_2,T,255, cv2.THRESH_BINARY_INV)
#Hiển thị kết quả:
plt.imshow(pic_inv,cmap='gray')
plt.xticks(range(0,img_2.shape[1],25))
plt.grid(ls='--')
plt.title('Ảnh căt ngưỡng - THRESH BINARY INV')
```

### Yêu	cầu	2.3.
Sử	dụng	phương	pháp	
phát	hiện	biên	Canny	với	các	tham	
số	phù	hợp	để	phát	hiện	đường	biên	
của	ảnh	như	dưới	đây:

```python
#Phát hiện biên Canny
img_canny1 = cv2.Canny(pic_inv, 80, 150)

#Hiển thị ảnh
plt.imshow(img_canny1, cmap='gray')
plt.title('Ảnh phát hiện biên với Canny')
plt.xticks(range(0,img_2.shape[1],25))
plt.grid(ls='--')
plt.show()
```

### Yêu	cầu	2.4.
Hiển	thị	các	kết	quả	thu	được	từ	yêu	cầu	1	đến	3	như	dưới	đây,	
lưu	ảnh	cắt	ngưỡng	và	ảnh	phát	hiện	cạnh	vào	thư	mục	Pic	đặt	tên	lần	lượt	là	
MSV_threshold.png	và	MSV_canny.png

```python
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.imshow(img_2,cmap='gray')
plt.title('Ảnh gốc ban đầu')
plt.xticks(range(0,img_2.shape[1],25))

plt.subplot(1,3,2)
plt.imshow(pic_inv,cmap='gray')
plt.title('Ảnh căt ngưỡng ( THRESH BINARY INV )')
plt.xticks(range(0,img_2.shape[1],25))


plt.subplot(1,3,3)
plt.imshow(img_canny1, cmap='gray')
plt.title('Ảnh phát hiện biên với Canny')
plt.xticks(range(0,img_2.shape[1],25))

plt.show()

cv2.imwrite('Pic/2221050156_threshold.png',pic_inv)
cv2.imwrite('Pic/2221050156_canny.png',img_canny1)
```

### BÀI	3:	
Sinh	viên	sử	dụng	thị	giác	máy	tính,	thiết	lập	các	tham	số	phù	hợp	để	thực	hiện	
đếm	số	quả	bóng	trong	hình	Pic_T3.jpg	

```python

#Bước 1: Đọc ảnh ở chế độ ảnh màu:
img_3 = cv2.imread('Pic/Pic_T3.jpg',cv2.IMREAD_COLOR_RGB)

#Hiển thị ảnh gốc ở dạng xám:
gray = cv2.cvtColor(img_3, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(12,10))
plt.subplot(2,2,1)
plt.imshow(gray, cmap='gray');
plt.title('Ảnh gốc')

#Bước 2: Phát hiện biên với Canny:
#Thực hiện làm mịn ảnh với Gausian:
blur = cv2.GaussianBlur(gray, (17,17), 0)

#Phát hiện biên với Canny
canny = cv2.Canny(blur, 60, 180, 3)
plt.subplot(2,2,2)
plt.imshow(canny, cmap='gray')
plt.title('Ảnh biên với Canny')


#Bước 3: Làm đậm đường biên:
dilated = cv2.dilate(canny, (1,1), iterations =3)
plt.subplot(2,2,3)
plt.imshow(dilated, cmap='gray')
plt.title('Ảnh làm đậm đường biên')

#Bước 4: Lấy countour của ảnh và đếm số lượng:
(cnt, heirarchy) = cv2.findContours(dilated.copy(), 
                                    cv2.RETR_EXTERNAL, 
                                    cv2.CHAIN_APPROX_NONE)

rgb = cv2.cvtColor(img_3, cv2.COLOR_BGR2RGB)
cv2.drawContours(rgb, cnt, -1, (500,0,0), 5)

plt.subplot(2,2,4)
plt.imshow(rgb)
plt.title('Ảnh chồng lấp countour')
plt.show()

print('Số lượng bong trong ảnh là: ', len(cnt))
```

### BÀI	4:	
Sinh	viên	đọc	ảnh	Pic_T4.jpg,	thực	hiện	tìm	kiếm	logo	HUMG	trong	ảnh.

```python
#Đọc và hiển thị ảnh màu: Pic_T4.jpg
img_4 = cv2.imread('Pic/Pic_T4.jpg',cv2.IMREAD_COLOR_RGB)
if img_4 is  None:
    print('đọc ảnh lỗi')
else:
    #hiển thị ảnh
    plt.imshow(img_4)
    plt.title('Ảnh gốc')
    #tăng mật độ của lưới :
    plt.xticks(range(0,img_4.shape[1],50),rotation=90)
    plt.yticks(range(0,img_4.shape[0],50))
    plt.grid(ls=':')#hiển thị lưới
    plt.show()
```

```python
#Cắt lấy logo
img_logo= img_4[1030:1155,1340:1460]

# hiển thị kết quả:
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.imshow(img_logo)
plt.title('Ảnh logo')
plt.show()
#lưu ảnh:
img_logo_rgb= cv2.cvtColor(img_logo, cv2.COLOR_BGR2RGB)
cv2.imwrite('Pic/2221050156_logo.png',img_logo_rgb)
```

```python

#img_4 = cv2.imread("Pic/Pic_T4.jpg", cv2.IMREAD_COLOR_RGB)
#Đọc ảnh mẫu:
temp1 = cv2.imread("Pic/2221050156_logo.png",cv2.IMREAD_COLOR_RGB)

#Hiển thị ảnh gốc - ảnh mẫu
plt.figure(figsize=(12,10))
plt.subplot(1,2,1)
plt.imshow(img_4,cmap='gray')
plt.title('Ảnh gốc')

plt.subplot(1,2,2)
plt.imshow(temp1,cmap='gray')
plt.title('Ảnh mẫu - Template')
plt.show()

       
```

```python
# Chuyển ảnh sang Xám để tăng tốc và ổn định
img_4_gray=cv2.cvtColor(img_4,cv2.COLOR_RGB2GRAY)
temp_gray=cv2.cvtColor(temp1,cv2.COLOR_RGB2GRAY)

#Sử dụng phương thức matchTemplate của OpenCV để tìm vùng phù hợp:
result=cv2.matchTemplate(img_4_gray,temp_gray,cv2.TM_CCOEFF)

#Xác định vị trí khớp tốt nhất:
min_val,max_val,min_loc,max_loc=cv2.minMaxLoc(result)
```

```python
#Hiển thị kết quả tìm kiếm:
#b1: lay kich thuoc cua anh mau
h,w=temp1.shape[:2]
print(h, w)
#b2 : lay toa do goc tren an trai cua vun khop tot nhat:
top_left=max_loc
bottom_right=top_left[0]+w,top_left[1]+h
#b3: ve hih chu nhat bao quan vung khop tot nhat tren anh goc
cv2.rectangle(img_4,top_left,bottom_right,(0,0,255),2)
#b4 hien thi ket qua 

plt.imshow(img_4)
plt.title('ket qua tim kiem anh mau  ')
plt.show()
```

```python
plt.figure(figsize=(15,5))
#ảnh kết quả
plt.subplot(1,2,1)
plt.imshow(img_4)
plt.title(' Kết quả tìm kiếm ảnh mẫu trên ảnh gốc')
plt.grid(ls='--')

#trực quan hóa tìm kiếm dạng heatmap:
plt.subplot(1,2,2)
plt.imshow(result,cmap='jet')
plt.title('Heatmap - Kết quả tìm kiếm')
#plt.colorbar()
plt.grid(ls='--')
plt.show()
```
